```python
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.6f}".format)
```

```python
tickers = ["AAPL", "MSFT", "AMZN", "JPM", "NVDA"]

prices = yf.download(
    tickers=tickers,
    start="2019-01-01",
    end="2025-12-31",
    auto_adjust=True,
    progress=False,
    threads=False
)

close_prices = prices["Close"].copy()

```
